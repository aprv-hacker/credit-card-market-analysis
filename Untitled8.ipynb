```python
def bar_perc(ax, col):
    total = len(data[col])  # Total count
    for p in ax.patches:
        percentage = '{:.1f}%'.format(100 * p.get_height() / total)
        x = p.get_x() + p.get_width() / 2
        y = p.get_height() + 0.01 * total  # Slight offset
        ax.text(x, y, percentage, ha='center', va='bottom', fontsize=6)
```

```python
payment_overview = pd.read_sql_query("""
SELECT 
    payment_type,
    COUNT(*) as transaction_count,
    COUNT(DISTINCT customer_id) as unique_customers,
    SUM(spend) as total_spend,
    AVG(spend) as avg_transaction_amount,
    MIN(spend) as min_spend,
    MAX(spend) as max_spend,
    COUNT(DISTINCT category) as categories_served,
    COUNT(DISTINCT month) as months_active
FROM fact_spends
GROUP BY payment_type
ORDER BY total_spend DESC
""", con)

print(" PAYMENT ECOSYSTEM OVERVIEW ")
payment_overview
```

```python
# Market share analysis
total_spend = payment_overview['total_spend'].sum()
total_transactions = payment_overview['transaction_count'].sum()

payment_overview['spend_market_share'] = (payment_overview['total_spend'] / total_spend * 100).round(2)
payment_overview['transaction_market_share'] = (payment_overview['transaction_count'] / total_transactions * 100).round(2)
payment_overview['customer_penetration'] = payment_overview['unique_customers'] / 4000 * 100  # Assuming 4000 total customers

print("\n=== MARKET SHARE ANALYSIS ===")
payment_overview[['payment_type', 'spend_market_share', 'transaction_market_share', 'customer_penetration']]
```

```python
payment_customer_df = pd.read_sql_query("""
WITH customer_totals AS (
    -- Calculate customer totals once
    SELECT 
        customer_id,
        SUM(spend) as customer_total_spend,
        COUNT(*) as customer_total_transactions
    FROM fact_spends
    GROUP BY customer_id
),
payment_summary AS (
    -- Calculate payment-level metrics
    SELECT 
        f.customer_id,

        d.age_group, d.city, d.occupation, d.gender, d."marital status", d.avg_income,
        f.payment_type,
        SUM(f.spend) as customer_payment_spend,
        COUNT(*) as customer_payment_transactions,
        AVG(f.spend) as customer_avg_transaction,
        COUNT(DISTINCT f.category) as categories_used,
        COUNT(DISTINCT f.month) as months_active_payment
        
    FROM fact_spends f
    JOIN dim_customers d ON f.customer_id = d.customer_id
    GROUP BY f.customer_id, f.payment_type, d.age_group, d.city, d.occupation, d.gender, d."marital status", d.avg_income
)

SELECT 
    ps.*,
    ct.customer_total_spend,
    ct.customer_total_transactions,
    ROUND(ps.customer_payment_spend * 100.0 / ct.customer_total_spend, 2) as payment_share_pct
FROM payment_summary ps
JOIN customer_totals ct ON ps.customer_id = ct.customer_id
""", con)
payment_customer_df
```

```python
# Payment method adoption by demographics
demo_payment_analysis = payment_customer_df.groupby(['payment_type', 'age_group', 'city', 'occupation']).agg({
    'customer_payment_spend': 'mean',
    'customer_id': 'count'
}).round(2)

print("\n=== PAYMENT ADOPTION BY DEMOGRAPHICS ===")
demo_payment_analysis.head(20)
```

```python
# Category spending patterns across all payment methods
category_payment_query = """
SELECT 
    category,
    payment_type,
    SUM(spend) as total_spend,
    COUNT(*) as transactions,
    COUNT(DISTINCT customer_id) as unique_customers,
    AVG(spend) as avg_spend_per_transaction,
    COUNT(DISTINCT month) as months_active
FROM fact_spends
GROUP BY category, payment_type
ORDER BY category, total_spend DESC
"""

category_payment_df = pd.read_sql_query(category_payment_query, con)

# Create comprehensive category-payment analysis
category_totals = category_payment_df.groupby('category')['total_spend'].sum()
category_payment_df['category_payment_share'] = category_payment_df.apply(
    lambda x: (x['total_spend'] / category_totals[x['category']] * 100).round(2), axis=1
)
```

```python
# Pivot for visualization
category_payment_pivot = category_payment_df.pivot_table(
    index='category',
    columns='payment_type',
    values='total_spend',
    fill_value=0
)

category_share_pivot = category_payment_df.pivot_table(
    index='category', 
    columns='payment_type',
    values='category_payment_share',
    fill_value=0
)

print("=== SPENDING BY CATEGORY AND PAYMENT METHOD ===")
category_payment_pivot
```

```python
print("\n=== MARKET SHARE BY CATEGORY ===")
category_share_pivot
```

```python
# Monthly trends for all payment methods
monthly_payment_query = """
SELECT 
    month,
    payment_type,
    SUM(spend) as monthly_spend,
    COUNT(*) as monthly_transactions,
    COUNT(DISTINCT customer_id) as active_customers,
    AVG(spend) as avg_transaction_size
FROM fact_spends
GROUP BY month, payment_type
ORDER BY 
    CASE month 
        WHEN 'May' THEN 1 
        WHEN 'June' THEN 2 
        WHEN 'July' THEN 3 
        WHEN 'August' THEN 4 
        WHEN 'September' THEN 5 
        WHEN 'October' THEN 6 
    END,
    monthly_spend DESC
"""

monthly_payment_df = pd.read_sql_query(monthly_payment_query, con)
```

```python
# Seasonal patterns
monthly_pivot = monthly_payment_df.pivot_table(
    index='month',
    columns='payment_type', 
    values='monthly_spend',
    fill_value=0
)

print("=== MONTHLY SPENDING PATTERNS BY PAYMENT METHOD ===")
monthly_pivot
```

```python
# Analyze customer payment behavior patterns - accounting for balanced transaction structure
def analyze_payment_behaviors_balanced(df):
    print("=== PAYMENT BEHAVIOR ANALYSIS (BALANCED DATA) ===")
    
    # Since all customers use all payment methods equally (same transaction counts),
    # focus on SPENDING INTENSITY rather than usage frequency
    
    customer_payment_profiles = df.pivot_table(
        index='customer_id',
        columns='payment_type',
        values='customer_payment_spend',  # Focus only on spend, not transactions
        fill_value=0,
        aggfunc='sum'
    )
    
    # Add demographic data
    customer_totals = df.groupby('customer_id').agg({
        'customer_total_spend': 'first',
        'customer_total_transactions': 'first', 
        'age_group': 'first',
        'city': 'first',
        'occupation': 'first',
        'avg_income': 'first'
    })
    
    customer_profiles = customer_payment_profiles.join(customer_totals)
    
    # Calculate payment method INTENSITY (spend share) 
    payment_methods = df['payment_type'].unique()
    for method in payment_methods:
        if method in customer_profiles.columns:
            customer_profiles[f'{method}_intensity'] = (customer_profiles[method] / 
                                                       customer_profiles['customer_total_spend'] * 100).fillna(0)
    
    # Payment behavior classification based on spending patterns
    def classify_payment_behavior(row):
        intensities = {}
        for method in payment_methods:
            intensity_col = f'{method}_intensity'
            if intensity_col in row:
                intensities[method] = row[intensity_col]
        
        # Find the payment method with highest spending intensity
        max_method = max(intensities, key=intensities.get) if intensities else 'Unknown'
        max_intensity = intensities.get(max_method, 0)
        
        # Classify based on intensity concentration
        if max_intensity > 45:
            return f"{max_method}_Dominant"
        elif max_intensity > 35:
            return f"{max_method}_Preferred"
        else:
            return "Balanced_User"
    
    customer_profiles['spending_behavior'] = customer_profiles.apply(classify_payment_behavior, axis=1)
    
    return customer_profiles

customer_payment_profiles = analyze_payment_behaviors_balanced(payment_customer_df)
print(f"Customer payment profiles shape: {customer_payment_profiles.shape}")

# Payment intensity insights
print("\n=== PAYMENT INTENSITY INSIGHTS ===")
print("Average payment method spending intensity per customer:")
payment_methods = payment_customer_df['payment_type'].unique()
for method in payment_methods:
    intensity_col = f'{method}_intensity'
    if intensity_col in customer_payment_profiles.columns:
        avg_intensity = customer_payment_profiles[intensity_col].mean()
        print(f"{method}: {avg_intensity:.1f}% spending intensity")

# Distribution of spending behaviors
spending_behavior_dist = customer_payment_profiles['spending_behavior'].value_counts()
print(f"\nCustomers by spending behavior pattern:")
print(spending_behavior_dist)

# Credit Card specific insights
if 'Credit Card_intensity' in customer_payment_profiles.columns:
    cc_intensity_stats = customer_payment_profiles['Credit Card_intensity'].describe()
    print(f"\nCredit Card Intensity Statistics:")
    print(f"  Mean: {cc_intensity_stats['mean']:.1f}%")
    print(f"  Std: {cc_intensity_stats['std']:.1f}%") 
    print(f"  Range: {cc_intensity_stats['min']:.1f}% - {cc_intensity_stats['max']:.1f}%")
    
    # Segment customers by CC intensity
    customer_payment_profiles['cc_intensity_segment'] = pd.cut(
        customer_payment_profiles['Credit Card_intensity'],
        bins=[0, 30, 40, 50, 100],
        labels=['Low_CC', 'Moderate_CC', 'High_CC', 'Very_High_CC']
    )
    
    cc_segment_dist = customer_payment_profiles['cc_intensity_segment'].value_counts()
    print(f"\nCustomers by CC Intensity Segments:")
    print(cc_segment_dist)
```

```python
def create_dominant_payment_method(df):
    # Find which payment method has highest spending for each customer
    payment_methods = ['Credit Card', 'Debit Card', 'UPI', 'Net Banking']  # Adjust based on your actual payment methods
    
    # Create intensity columns if they don't exist
    payment_intensities = {}
    for method in payment_methods:
        if f'{method}_intensity' in df.columns:
            payment_intensities[method] = f'{method}_intensity'
        elif method in df.columns:  # If direct spending columns exist
            payment_intensities[method] = method
    
    if payment_intensities:
        # Find the payment method with maximum intensity for each customer
        intensity_cols = list(payment_intensities.values())
        df['dominant_payment_method'] = df[intensity_cols].idxmax(axis=1)
        
        # Clean up column names (remove '_intensity' suffix if present)
        df['dominant_payment_method'] = df['dominant_payment_method'].str.replace('_intensity', '')
    else:
        print("No payment method columns found")
        return df
    
    return df

# Create the missing column
customer_profiles_segmented = create_dominant_payment_method(customer_profiles_segmented)
```

```python
# Since CC is already #1 with 40.74% market share, focus on enhancement opportunities
def analyze_cc_enhancement_opportunities(df):
    # Create customer-level payment profiles
    customer_payment_profiles = df.pivot_table(
        index='customer_id',
        columns='payment_type',
        values=['customer_payment_spend', 'customer_payment_transactions'],
        fill_value=0,
        aggfunc='sum'
    )
    
    # Flatten columns
    customer_payment_profiles.columns = ['_'.join(col).strip() for col in customer_payment_profiles.columns]
    
    # Add demographics and totals
    customer_totals = df.groupby('customer_id').agg({
        'customer_total_spend': 'first',
        'customer_total_transactions': 'first', 
        'age_group': 'first',
        'city': 'first',
        'occupation': 'first',
        'avg_income': 'first'
    })
    
    customer_profiles = customer_payment_profiles.join(customer_totals)
    
    # Calculate payment shares (all customers use all payment methods)
    payment_methods = df['payment_type'].unique()
    for method in payment_methods:
        spend_col = f'customer_payment_spend_{method}'
        if spend_col in customer_profiles.columns:
            customer_profiles[f'{method}_share'] = (customer_profiles[spend_col] / 
                                                   customer_profiles['customer_total_spend'] * 100).fillna(0)
    
    # CC Enhancement Segmentation
    cc_share_col = 'Credit Card_share'
    if cc_share_col in customer_profiles.columns:
        # Segment by CC usage intensity and enhancement potential
        cc_quartiles = customer_profiles[cc_share_col].quantile([0.25, 0.5, 0.75])
        
        def categorize_cc_enhancement(row):
            cc_share = row[cc_share_col]
            income = row['avg_income']
            total_spend = row['customer_total_spend']
            
            if cc_share >= cc_quartiles[0.75]:
                return 'CC Champions'  # High CC users - loyalty programs
            elif cc_share >= cc_quartiles[0.5]:
                if income > customer_profiles['avg_income'].median():
                    return 'Premium Expansion'  # Medium CC + High income - upgrade opportunities
                else:
                    return 'Mainstream Users'  # Medium CC + Medium income - retention focus
            else:
                if total_spend > customer_profiles['customer_total_spend'].median():
                    return 'High Potential'  # Low CC + High spend - conversion opportunity
                else:
                    return 'Basic Tier'  # Low CC + Low spend - basic products
        
        customer_profiles['cc_enhancement_segment'] = customer_profiles.apply(categorize_cc_enhancement, axis=1)
    
    return customer_profiles

customer_profiles_enhanced = analyze_cc_enhancement_opportunities(payment_customer_df)
print(f"Customer profiles with CC enhancement segments: {customer_profiles_enhanced.shape}")

# Analysis of CC enhancement opportunities
print("\n=== CREDIT CARD ENHANCEMENT OPPORTUNITIES ===")
if 'Credit Card_share' in customer_profiles_enhanced.columns:
    cc_stats = customer_profiles_enhanced['Credit Card_share'].describe()
    print(f"CC Share Statistics:")
    print(f"  Mean: {cc_stats['mean']:.1f}%")
    print(f"  Range: {cc_stats['min']:.1f}% - {cc_stats['max']:.1f}%")
    
    enhancement_segments = customer_profiles_enhanced['cc_enhancement_segment'].value_counts()
    print(f"\nCC Enhancement Segments:")
    for segment, count in enhancement_segments.items():
        print(f"  {segment}: {count} customers ({count/len(customer_profiles_enhanced)*100:.1f}%)")

# Opportunity analysis by demographics
demo_cc_analysis = customer_profiles_enhanced.groupby(['age_group', 'city', 'occupation']).agg({
    'Credit Card_share': 'mean',
    'customer_total_spend': 'mean',
    'avg_income': 'mean'
}).round(2)

print("\n=== CC SHARE BY DEMOGRAPHICS (Top Opportunities) ===")
low_cc_high_potential = demo_cc_analysis[
    (demo_cc_analysis['Credit Card_share'] < 35) &  # Below average CC share
    (demo_cc_analysis['customer_total_spend'] > demo_cc_analysis['customer_total_spend'].median())
].sort_values('customer_total_spend', ascending=False)

print("Demographics with low CC share but high spending potential:")
print(low_cc_high_potential.head(10))
```

```python
# Demographic and behavioral clustering for meaningful customer segments
def perform_demographic_behavioral_clustering(df):
    
    # Get category spending patterns for each customer
    category_spending_query = """
    SELECT 
        customer_id,
        category,
        SUM(spend) as category_total_spend
    FROM fact_spends
    GROUP BY customer_id, category
    """
    
    category_customer_spend = pd.read_sql_query(category_spending_query, con)
    
    # Pivot to get category spending distribution per customer
    category_pivot = category_customer_spend.pivot_table(
        index='customer_id', 
        columns='category', 
        values='category_total_spend', 
        fill_value=0
    )
    
    print(f"Category spending patterns shape: {category_pivot.shape}")
    print("Categories analyzed:", category_pivot.columns.tolist())
    
    # Core clustering features: Demographics + Income + Spending + Category Patterns
    clustering_features = [
        'customer_total_spend', 'avg_income', 'customer_total_transactions'
    ]
    
    # Add demographic dummy variables  
    demo_dummies = pd.get_dummies(df[['age_group', 'city', 'occupation']], prefix=['age', 'city', 'occ'])
    
    # Combine all features
    X_cluster = pd.concat([
        df[clustering_features],
        category_pivot,  # Category spending patterns - THIS IS KEY
        demo_dummies     # Demographics
    ], axis=1).fillna(0)
    
    print(f"Total clustering features: {X_cluster.shape[1]}")
    
    # Standardize features
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_cluster)
    
    # Find optimal clusters
    silhouette_scores = []
    K_range = range(2, 11)
    
    for k in K_range:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        labels = kmeans.fit_predict(X_scaled)
        silhouette_scores.append(silhouette_score(X_scaled, labels))
    
    # Plot silhouette scores
    plt.figure(figsize=(10, 6))
    plt.plot(K_range, silhouette_scores, 'bo-')
    plt.title('Optimal Customer Clusters (Demographics + Category Spending)')
    plt.xlabel('Number of Clusters')
    plt.ylabel('Silhouette Score')
    plt.show()
    
    # Apply optimal clustering
    optimal_k = K_range[np.argmax(silhouette_scores)]
    print(f"Optimal number of clusters: {optimal_k}")
    
    kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
    df['customer_cluster'] = kmeans_final.fit_predict(X_scaled)
    
    # Merge category data with customer profiles for analysis
    df_with_categories = df.join(category_pivot, how='left')
    
    # Analyze clusters with category insights
    cluster_analysis = df_with_categories.groupby('customer_cluster').agg({
        'customer_total_spend': ['count', 'mean'],
        'avg_income': 'mean',
        'Credit Card_share': 'mean',
        'age_group': lambda x: x.mode().iloc[0],
        'city': lambda x: x.mode().iloc[0], 
        'occupation': lambda x: x.mode().iloc[0],
        # Top spending categories per cluster
        'Electronics': 'mean',
        'Groceries': 'mean', 
        'Travel': 'mean',
        'Food': 'mean',
        'Health & Wellness': 'mean',
        'Entertainment': 'mean'
    }).round(2)
    
    # Identify top category for each cluster
    category_cols = ['Electronics', 'Groceries', 'Travel', 'Food', 'Health & Wellness', 'Entertainment']
    cluster_top_categories = {}
    
    for cluster in df_with_categories['customer_cluster'].unique():
        cluster_data = df_with_categories[df_with_categories['customer_cluster'] == cluster]
        top_cats = cluster_data[category_cols].mean().nlargest(3)
        cluster_top_categories[cluster] = {
            'top_categories': top_cats.index.tolist(),
            'spending_amounts': top_cats.values.tolist()
        }
    
    return df_with_categories, cluster_analysis, cluster_top_categories, kmeans_final, scaler

customer_profiles_clustered, cluster_analysis, cluster_categories, clustering_model, feature_scaler = perform_demographic_behavioral_clustering(customer_profiles_enhanced)

print("=== DEMOGRAPHIC & BEHAVIORAL CLUSTERING ANALYSIS ===")
print(cluster_analysis)

print("\n=== TOP CATEGORIES BY CLUSTER ===")
for cluster, details in cluster_categories.items():
    print(f"Cluster {cluster}: {details['top_categories']} (avg spend: ₹{details['spending_amounts'][0]:,.0f}, ₹{details['spending_amounts'][1]:,.0f}, ₹{details['spending_amounts'][2]:,.0f})")

# Visualize clusters with better context
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Income vs Total Spend by cluster
scatter1 = axes[0].scatter(customer_profiles_clustered['avg_income'], 
                          customer_profiles_clustered['customer_total_spend'],
                          c=customer_profiles_clustered['customer_cluster'], 
                          cmap='viridis', alpha=0.6)
axes[0].set_xlabel('Average Income (₹)')
axes[0].set_ylabel('Total Spend (₹)')
axes[0].set_title('Clusters: Income vs Spending')

# Age distribution by cluster
age_cluster = pd.crosstab(customer_profiles_clustered['age_group'], customer_profiles_clustered['customer_cluster'])
age_cluster.plot(kind='bar', ax=axes[1], stacked=True)
axes[1].set_title('Age Distribution by Cluster')
axes[1].tick_params(axis='x', rotation=45)

# City distribution by cluster
city_cluster = pd.crosstab(customer_profiles_clustered['city'], customer_profiles_clustered['customer_cluster'])
city_cluster.plot(kind='bar', ax=axes[2], stacked=True)
axes[2].set_title('City Distribution by Cluster') 
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Analyze category preferences by customer segments for reward program design
category_spend_query = """
SELECT 
    customer_id,
    category,
    payment_type,
    SUM(spend) as category_spend,
    COUNT(*) as category_transactions
FROM fact_spends
GROUP BY customer_id, category, payment_type
"""

category_customer_df = pd.read_sql_query(category_spend_query, con)

# Merge with customer profiles
category_analysis = category_customer_df.merge(
    customer_profiles_clustered[['customer_cluster', 'cc_enhancement_segment', 'age_group', 'city', 'avg_income']], 
    left_on='customer_id', right_index=True
)

# Category preferences by cluster
def analyze_category_preferences(df):
    # Category spending by customer cluster
    cluster_category = df.groupby(['customer_cluster', 'category']).agg({
        'category_spend': 'sum',
        'customer_id': 'nunique'
    }).reset_index()
    
    # Category preferences by CC enhancement segment
    segment_category = df.groupby(['cc_enhancement_segment', 'category']).agg({
        'category_spend': 'sum',
        'customer_id': 'nunique'
    }).reset_index()
    
    # Top categories for each cluster
    top_categories_by_cluster = {}
    for cluster in df['customer_cluster'].unique():
        cluster_data = cluster_category[cluster_category['customer_cluster'] == cluster]
        top_cats = cluster_data.nlargest(3, 'category_spend')['category'].tolist()
        top_categories_by_cluster[cluster] = top_cats
    
    # Reward category recommendations
    reward_recommendations = {}
    for segment in df['cc_enhancement_segment'].unique():
        segment_data = segment_category[segment_category['cc_enhancement_segment'] == segment]
        top_reward_cats = segment_data.nlargest(3, 'category_spend')['category'].tolist()
        total_spend = segment_data['category_spend'].sum()
        
        reward_recommendations[segment] = {
            'primary_categories': top_reward_cats,
            'total_spend_potential': total_spend,
            'customer_count': segment_data['customer_id'].sum()
        }
    
    return top_categories_by_cluster, reward_recommendations

category_by_cluster, reward_strategy = analyze_category_preferences(category_analysis)

print("=== CATEGORY PREFERENCES BY CUSTOMER CLUSTER ===")
for cluster, categories in category_by_cluster.items():
    print(f"Cluster {cluster}: {categories}")

print("\n=== REWARD CATEGORY RECOMMENDATIONS ===")
for segment, details in reward_strategy.items():
    print(f"\n{segment}:")
    print(f"  Recommended reward categories: {details['primary_categories']}")
    print(f"  Total spend potential: ₹{details['total_spend_potential']:,.0f}")
    print(f"  Target customer count: {details['customer_count']}")

# Visualize category preferences
category_pivot = category_analysis.groupby(['customer_cluster', 'category'])['category_spend'].sum().unstack(fill_value=0)

plt.figure(figsize=(14, 8))
sns.heatmap(category_pivot, annot=True, fmt='.0f', cmap='YlOrRd')
plt.title('Category Spending Patterns by Customer Cluster')
plt.xlabel('Spending Categories')
plt.ylabel('Customer Clusters')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

# Connect to database (assuming you have this)
con = sqlite3.connect('inventory.db')

# ==================== ENHANCED CLUSTERING FUNCTION ====================

def perform_demographic_behavioral_clustering(df):
    """
    Enhanced clustering with feature engineering, PCA, and algorithm comparison
    Addresses the 0.24 silhouette score issue
    """
    
    print("="*70)
    print("ENHANCED DEMOGRAPHIC & BEHAVIORAL CLUSTERING")
    print("="*70)
    
    # ==================== DATA PREPARATION ====================
    
    # Get category spending patterns for each customer
    category_spending_query = """
    SELECT 
        customer_id,
        category,
        SUM(spend) as category_total_spend
    FROM fact_spends
    GROUP BY customer_id, category
    """
    
    category_customer_spend = pd.read_sql_query(category_spending_query, con)
    
    # Pivot to get category spending distribution per customer
    category_pivot = category_customer_spend.pivot_table(
        index='customer_id', 
        columns='category', 
        values='category_total_spend', 
        fill_value=0
    )
    
    print(f"\nCategory spending patterns shape: {category_pivot.shape}")
    print("Categories analyzed:", category_pivot.columns.tolist())
    
    # ==================== FEATURE ENGINEERING ====================
    
    print("\n--- Feature Engineering ---")
    
    # 1. Create ratio features (more meaningful than raw values)
    df['spend_to_income_ratio'] = df['customer_total_spend'] / (df['avg_income'] + 1)
    df['spend_per_transaction'] = df['customer_total_spend'] / (df['customer_total_transactions'] + 1)
    
    print("✓ Added ratio features: spend_to_income_ratio, spend_per_transaction")
    
    # 2. Category spending as PERCENTAGES (normalizes across different spending levels)
    category_percentages = category_pivot.div(category_pivot.sum(axis=1), axis=0).fillna(0)
    category_percentages.columns = [f"{col}_pct" for col in category_percentages.columns]
    
    print(f"✓ Converted category spending to percentages: {category_percentages.shape[1]} features")
    
    # 3. Core clustering features
    clustering_features = [
        'customer_total_spend', 
        'avg_income', 
        'customer_total_transactions',
        'spend_to_income_ratio',
        'spend_per_transaction'
    ]
    
    # Add Credit Card share if it exists
    if 'Credit Card_share' in df.columns:
        clustering_features.append('Credit Card_share')
        print("✓ Included Credit Card share in features")
    
    # 4. Demographic dummy variables (drop_first to avoid multicollinearity)
    demo_dummies = pd.get_dummies(
        df[['age_group', 'city', 'occupation']], 
        prefix=['age', 'city', 'occ'],
        drop_first=True
    )
    
    print(f"✓ Created demographic dummies: {demo_dummies.shape[1]} features")
    
    # ==================== COMBINE FEATURES ====================
    
    # Combine all features
    X_cluster = pd.concat([
        df[clustering_features],
        category_percentages,  # Using percentages instead of absolute values
        demo_dummies
    ], axis=1).fillna(0)
    
    print(f"\nTotal features before PCA: {X_cluster.shape[1]}")
    
    # ==================== SCALING ====================
    
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_cluster)
    
    print("✓ Features scaled using StandardScaler")
    
    # ==================== DIMENSIONALITY REDUCTION (PCA) ====================
    
    print("\n--- Applying PCA ---")
    
    pca = PCA(n_components=0.95, random_state=42)  # Keep 95% of variance
    X_pca = pca.fit_transform(X_scaled)
    
    explained_variance = pca.explained_variance_ratio_.sum()
    print(f"✓ Reduced from {X_scaled.shape[1]} to {X_pca.shape[1]} features")
    print(f"✓ Retained {explained_variance:.1%} of variance")
    
    # ==================== ALGORITHM COMPARISON ====================
    
    print("\n--- Testing Multiple Clustering Algorithms ---")
    
    results = {}
    K_range = range(3, 8)  # Business-friendly range: 3-7 clusters
    
    print(f"Testing k={min(K_range)} to k={max(K_range)} clusters...")
    
    for k in K_range:
        # K-Means
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=20, max_iter=300)
        labels_km = kmeans.fit_predict(X_pca)
        
        # Gaussian Mixture Model
        gmm = GaussianMixture(n_components=k, random_state=42, 
                             covariance_type='full', max_iter=200)
        labels_gmm = gmm.fit_predict(X_pca)
        
        # Store results
        results[f'KMeans_k{k}'] = {
            'algorithm': 'K-Means',
            'k': k,
            'labels': labels_km,
            'silhouette': silhouette_score(X_pca, labels_km),
            'davies_bouldin': davies_bouldin_score(X_pca, labels_km),
            'calinski_harabasz': calinski_harabasz_score(X_pca, labels_km),
            'inertia': kmeans.inertia_
        }
        
        results[f'GMM_k{k}'] = {
            'algorithm': 'GMM',
            'k': k,
            'labels': labels_gmm,
            'silhouette': silhouette_score(X_pca, labels_gmm),
            'davies_bouldin': davies_bouldin_score(X_pca, labels_gmm),
            'calinski_harabasz': calinski_harabasz_score(X_pca, labels_gmm),
            'bic': gmm.bic(X_pca),
            'aic': gmm.aic(X_pca)
        }
    
    # ==================== SELECT BEST MODEL ====================
    
    # Create comparison table
    comparison_data = []
    for key, res in results.items():
        comparison_data.append({
            'Algorithm': res['algorithm'],
            'K': res['k'],
            'Silhouette': res['silhouette'],
            'Davies-Bouldin': res['davies_bouldin'],
            'Calinski-Harabasz': res['calinski_harabasz']
        })
    
    comparison_df = pd.DataFrame(comparison_data)
    comparison_df = comparison_df.sort_values('Silhouette', ascending=False)
    
    print("\n" + "="*70)
    print("CLUSTERING RESULTS (Sorted by Silhouette Score)")
    print("="*70)
    print(comparison_df.to_string(index=False))
    
    # Select best model based on silhouette score
    best_key = max(results, key=lambda x: results[x]['silhouette'])
    best_result = results[best_key]
    
    print(f"\n{'='*70}")
    print(f"BEST MODEL SELECTED")
    print(f"{'='*70}")
    print(f"Algorithm: {best_result['algorithm']}")
    print(f"Number of Clusters: {best_result['k']}")
    print(f"Silhouette Score: {best_result['silhouette']:.3f}")
    print(f"Davies-Bouldin Index: {best_result['davies_bouldin']:.3f}")
    print(f"Calinski-Harabasz Score: {best_result['calinski_harabasz']:,.0f}")
    
    # Interpretation
    silhouette = best_result['silhouette']
    if silhouette > 0.5:
        interpretation = "EXCELLENT: Strong, well-separated clusters"
    elif silhouette > 0.35:
        interpretation = "GOOD: Reasonable cluster structure for customer data"
    elif silhouette > 0.25:
        interpretation = "ACCEPTABLE: Weak but usable clusters (typical for customer segmentation)"
    else:
        interpretation = "POOR: No substantial cluster structure"
    
    print(f"\nInterpretation: {interpretation}")
    
    # Apply best clustering to dataframe
    df['customer_cluster'] = best_result['labels']
    
    # ==================== CLUSTER ANALYSIS ====================
    
    print(f"\n{'='*70}")
    print("CLUSTER PROFILES")
    print(f"{'='*70}")
    
    # Merge category data with customer profiles
    df_with_categories = df.join(category_pivot, how='left')
    
    # Get all category columns
    category_cols = [col for col in category_pivot.columns]
    
    # Aggregate cluster statistics
    cluster_agg = {
        'customer_total_spend': ['count', 'mean'],
        'avg_income': 'mean',
        'customer_total_transactions': 'mean',
        'spend_to_income_ratio': 'mean',
        'spend_per_transaction': 'mean'
    }
    
    # Add Credit Card share if exists
    if 'Credit Card_share' in df.columns:
        cluster_agg['Credit Card_share'] = 'mean'
    
    # Add demographics
    cluster_agg.update({
        'age_group': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'Unknown',
        'city': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'Unknown',
        'occupation': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'Unknown'
    })
    
    # Add top categories (first 6 available)
    available_cats = [cat for cat in category_cols if cat in df_with_categories.columns][:6]
    for cat in available_cats:
        cluster_agg[cat] = 'mean'
    
    cluster_analysis = df_with_categories.groupby('customer_cluster').agg(cluster_agg).round(2)
    
    print(cluster_analysis)
    
    # ==================== TOP CATEGORIES BY CLUSTER ====================
    
    print(f"\n{'='*70}")
    print("TOP SPENDING CATEGORIES BY CLUSTER")
    print(f"{'='*70}")
    
    cluster_top_categories = {}
    
    for cluster in sorted(df_with_categories['customer_cluster'].unique()):
        cluster_data = df_with_categories[df_with_categories['customer_cluster'] == cluster]
        
        # Calculate top categories for this cluster
        if len(category_cols) > 0:
            available = [col for col in category_cols if col in cluster_data.columns]
            if available:
                top_cats = cluster_data[available].mean().nlargest(3)
                
                cluster_top_categories[cluster] = {
                    'top_categories': top_cats.index.tolist(),
                    'spending_amounts': top_cats.values.tolist()
                }
                
                print(f"\nCluster {cluster}:")
                for i, (cat, amt) in enumerate(zip(top_cats.index, top_cats.values), 1):
                    print(f"  {i}. {cat}: ₹{amt:,.0f}")
    
    # ==================== VISUALIZATIONS ====================
    
    print(f"\n{'='*70}")
    print("GENERATING VISUALIZATIONS")
    print(f"{'='*70}")
    
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    # 1. Income vs Total Spend by cluster
    scatter1 = axes[0].scatter(df_with_categories['avg_income'], 
                              df_with_categories['customer_total_spend'],
                              c=df_with_categories['customer_cluster'], 
                              cmap='viridis', alpha=0.6)
    axes[0].set_xlabel('Average Income (₹)')
    axes[0].set_ylabel('Total Spend (₹)')
    axes[0].set_title('Clusters: Income vs Spending')
    plt.colorbar(scatter1, ax=axes[0])
    
    # 2. Age distribution by cluster
    age_cluster = pd.crosstab(df_with_categories['age_group'], df_with_categories['customer_cluster'])
    age_cluster.plot(kind='bar', ax=axes[1], stacked=True)
    axes[1].set_title('Age Distribution by Cluster')
    axes[1].tick_params(axis='x', rotation=45)
    axes[1].set_xlabel('Age Group')
    axes[1].set_ylabel('Number of Customers')
    
    # 3. City distribution by cluster
    city_cluster = pd.crosstab(df_with_categories['city'], df_with_categories['customer_cluster'])
    city_cluster.plot(kind='bar', ax=axes[2], stacked=True)
    axes[2].set_title('City Distribution by Cluster') 
    axes[2].tick_params(axis='x', rotation=45)
    axes[2].set_xlabel('City')
    axes[2].set_ylabel('Number of Customers')
    
    plt.tight_layout()
    plt.show()
    
    # Category heatmap
    if len(available_cats) > 0:
        plt.figure(figsize=(12, 8))
        cluster_category_matrix = df_with_categories.groupby('customer_cluster')[available_cats].mean()
        
        sns.heatmap(cluster_category_matrix, annot=True, fmt='.0f', cmap='YlOrRd',
                   xticklabels=available_cats,
                   yticklabels=[f'Cluster {i}' for i in cluster_category_matrix.index])
        plt.title('Average Category Spending by Customer Cluster (₹)')
        plt.xlabel('Spending Categories')
        plt.ylabel('Customer Clusters')
        plt.tight_layout()
        plt.show()
    
    print(f"\n{'='*70}")
    print("CLUSTERING COMPLETE")
    print(f"{'='*70}")
    
    return df_with_categories, cluster_analysis, cluster_top_categories, best_result, scaler

# ==================== EXECUTE CLUSTERING ====================

# Run the enhanced clustering function
customer_profiles_clustered, cluster_analysis, cluster_categories, clustering_model, feature_scaler = perform_demographic_behavioral_clustering(customer_profiles_enhanced)

# ==================== DISPLAY RESULTS ====================

print("\n" + "="*70)
print("=== DEMOGRAPHIC & BEHAVIORAL CLUSTERING ANALYSIS ===")
print("="*70)
print(cluster_analysis)

print("\n" + "="*70)
print("=== TOP CATEGORIES BY CLUSTER ===")
print("="*70)
for cluster, details in cluster_categories.items():
    amounts = details['spending_amounts']
    print(f"Cluster {cluster}: {details['top_categories']} (avg spend: ₹{amounts[0]:,.0f}, ₹{amounts[1]:,.0f}, ₹{amounts[2]:,.0f})")

print("\n" + "="*70)
print("IMPROVEMENTS SUMMARY")
print("="*70)
```

```python
def analyze_category_reward_opportunities_twostage(df):
    """
    Analyze category spending patterns for each of the 8 clusters
    and create reward program recommendations
    """
    
    print("="*70)
    print("STEP 8: CATEGORY SPENDING ANALYSIS & REWARD RECOMMENDATIONS")
    print("="*70)
    
    # Get category spending for all customers
    category_spending_query = """
    SELECT 
        customer_id,
        category,
        SUM(spend) as category_total_spend
    FROM fact_spends
    GROUP BY customer_id, category
    """
    
    category_customer_spend = pd.read_sql_query(category_spending_query, con)
    category_pivot = category_customer_spend.pivot_table(
        index='customer_id', 
        columns='category', 
        values='category_total_spend', 
        fill_value=0
    )
    
    # Merge with clustered data
    df_with_categories = df.join(category_pivot, how='left')
    
    # Get available category columns
    category_cols = [col for col in category_pivot.columns]
    print(f"\nCategories available: {category_cols}")
    
    # Analyze each cluster
    reward_recommendations = {}
    
    for cluster in sorted(df['final_cluster'].unique()):
        cluster_data = df_with_categories[df_with_categories['final_cluster'] == cluster]
        
        # Skip if cluster too small
        if len(cluster_data) < 10:
            continue
        
        # Calculate top categories for this cluster
        available_cats = [col for col in category_cols if col in cluster_data.columns]
        if not available_cats:
            continue
            
        top_cats = cluster_data[available_cats].mean().nlargest(3)
        
        # Demographics
        primary_age = cluster_data['age_group'].mode().iloc[0] if 'age_group' in cluster_data.columns else 'Unknown'
        primary_city = cluster_data['city'].mode().iloc[0] if 'city' in cluster_data.columns else 'Unknown'
        primary_occupation = cluster_data['occupation'].mode().iloc[0] if 'occupation' in cluster_data.columns else 'Unknown'
        
        # Financial metrics
        avg_income = cluster_data['avg_income'].mean() if 'avg_income' in cluster_data.columns else 0
        avg_spend = cluster_data['customer_total_spend'].mean() if 'customer_total_spend' in cluster_data.columns else 0
        avg_cc_share = cluster_data['Credit Card_share'].mean() if 'Credit Card_share' in cluster_data.columns else 0
        
        # Reward potential (2% on top categories)
        total_category_spend = top_cats.sum()
        reward_potential = total_category_spend * 0.02
        
        reward_recommendations[f'Cluster_{cluster}'] = {
            'cluster_id': cluster,
            'size': len(cluster_data),
            'primary_demographics': f"{primary_age} {primary_occupation}s in {primary_city}",
            'avg_income': f"₹{avg_income:,.0f}",
            'avg_spend': f"₹{avg_spend:,.0f}",
            'cc_share': f"{avg_cc_share:.1f}%",
            'top_categories': top_cats.index.tolist(),
            'category_spending': [f"₹{amt:,.0f}" for amt in top_cats.values],
            'reward_potential_per_customer': f"₹{reward_potential:,.0f}",
            'cluster_reward_opportunity': f"₹{reward_potential * len(cluster_data):,.0f}"
        }
    
    # Display results
    print("\n" + "="*70)
    print("REWARD PROGRAM RECOMMENDATIONS BY CLUSTER")
    print("="*70)
    
    for cluster_name, rec in reward_recommendations.items():
        print(f"\n{cluster_name} - {rec['primary_demographics']}")
        print(f"  Size: {rec['size']:,} customers")
        print(f"  Income: {rec['avg_income']}, Spend: {rec['avg_spend']}, CC Share: {rec['cc_share']}")
        print(f"  Top Categories: {', '.join(rec['top_categories'][:3])}")
        print(f"  Category Spending: {', '.join(rec['category_spending'][:3])}")
        print(f"  Reward Potential: {rec['reward_potential_per_customer']}/customer")
        print(f"  Total Opportunity: {rec['cluster_reward_opportunity']}")
    
    # Visualization: Category spending heatmap
    cluster_category_matrix = df_with_categories.groupby('final_cluster')[available_cats].mean()
    
    plt.figure(figsize=(14, 10))
    sns.heatmap(cluster_category_matrix, annot=True, fmt='.0f', cmap='YlOrRd',
               xticklabels=available_cats,
               yticklabels=[f'Cluster {i}' for i in cluster_category_matrix.index])
    plt.title('Average Category Spending by Customer Cluster (₹)', fontsize=14, fontweight='bold')
    plt.xlabel('Spending Categories', fontsize=12)
    plt.ylabel('Customer Clusters', fontsize=12)
    plt.tight_layout()
    plt.show()
    
    return reward_recommendations, df_with_categories

# Run Step 8
reward_strategy, customer_profiles_with_categories = analyze_category_reward_opportunities_twostage(customer_profiles_clustered)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score
import sqlite3
import warnings
warnings.filterwarnings('ignore')

# Connect to database
con = sqlite3.connect('inventory.db')

# Load customer profiles
print("Loading customer data...")
customer_query = """
SELECT 
    c.*,
    SUM(f.spend) as customer_total_spend,
    COUNT(*) as customer_total_transactions
FROM dim_customers c
JOIN fact_spends f ON c.customer_id = f.customer_id
GROUP BY c.customer_id
"""
customer_profiles = pd.read_sql_query(customer_query, con)
print(f"Loaded {len(customer_profiles)} customers")

# Get category spending
print("Loading category spending data...")
category_query = """
SELECT 
    customer_id,
    category,
    SUM(spend) as category_total_spend
FROM fact_spends
GROUP BY customer_id, category
"""
category_customer_spend = pd.read_sql_query(category_query, con)

category_pivot = category_customer_spend.pivot_table(
    index='customer_id',
    columns='category',
    values='category_total_spend',
    fill_value=0
)

category_percentages = category_pivot.div(category_pivot.sum(axis=1), axis=0).fillna(0)
print(f"Categories available: {category_pivot.columns.tolist()}")

# Prepare features for clustering
clustering_features = ['customer_total_spend', 'avg_income', 'customer_total_transactions']

# Combine with category percentages
available_cats = category_percentages.columns.tolist()[:5]
all_features = pd.concat([
    customer_profiles.set_index('customer_id')[clustering_features],
    category_percentages[available_cats]
], axis=1).fillna(0)

print(f"\nTotal features for clustering: {all_features.shape[1]}")
print(f"Features: {all_features.columns.tolist()}")

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(all_features)

print("\n" + "="*70)
print("GAUSSIAN MIXTURE MODEL (GMM) CLUSTERING")
print("="*70)

# Test different numbers of components
bic_scores = []
aic_scores = []
silhouette_scores_gmm = []
n_components_range = range(3, 11)

print("\nTesting different numbers of components...")
for n in n_components_range:
    gmm = GaussianMixture(n_components=n, random_state=42, n_init=10)
    gmm.fit(X_scaled)
    
    labels = gmm.predict(X_scaled)
    
    bic = gmm.bic(X_scaled)
    aic = gmm.aic(X_scaled)
    sil = silhouette_score(X_scaled, labels)
    
    bic_scores.append(bic)
    aic_scores.append(aic)
    silhouette_scores_gmm.append(sil)
    
    print(f"n={n}: BIC={bic:,.0f}, AIC={aic:,.0f}, Silhouette={sil:.3f}")

# Find optimal number of components
optimal_n = n_components_range[np.argmin(bic_scores)]
print(f"\nOptimal number of components: {optimal_n} (based on BIC)")

# Fit final GMM model
gmm_final = GaussianMixture(n_components=optimal_n, random_state=42, n_init=10)
gmm_final.fit(X_scaled)

# Get cluster assignments and probabilities
customer_profiles['gmm_cluster'] = gmm_final.predict(X_scaled)
cluster_probabilities = gmm_final.predict_proba(X_scaled)
customer_profiles['gmm_confidence'] = cluster_probabilities.max(axis=1)

print(f"\nGMM Clustering Results:")
print(f"Silhouette Score: {silhouette_score(X_scaled, customer_profiles['gmm_cluster']):.3f}")
print(f"Customers with low confidence (<70%): {(customer_profiles['gmm_confidence'] < 0.7).sum()}")
print(f"Average confidence: {customer_profiles['gmm_confidence'].mean():.3f}")

# Visualize GMM selection metrics
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(n_components_range, bic_scores, 'bo-', linewidth=2, markersize=8)
axes[0].axvline(optimal_n, color='red', linestyle='--', label=f'Optimal n={optimal_n}')
axes[0].set_xlabel('Number of Components')
axes[0].set_ylabel('BIC Score')
axes[0].set_title('BIC Score (Lower is Better)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(n_components_range, aic_scores, 'go-', linewidth=2, markersize=8)
axes[1].axvline(optimal_n, color='red', linestyle='--', label=f'Optimal n={optimal_n}')
axes[1].set_xlabel('Number of Components')
axes[1].set_ylabel('AIC Score')
axes[1].set_title('AIC Score (Lower is Better)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

axes[2].plot(n_components_range, silhouette_scores_gmm, 'ro-', linewidth=2, markersize=8)
axes[2].axvline(optimal_n, color='red', linestyle='--', label=f'Optimal n={optimal_n}')
axes[2].set_xlabel('Number of Components')
axes[2].set_ylabel('Silhouette Score')
axes[2].set_title('Silhouette Score (Higher is Better)')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('gmm_component_selection.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n" + "="*70)
print("t-SNE VISUALIZATION")
print("="*70)

# Apply t-SNE
print("\nApplying t-SNE (this may take a few minutes)...")
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
X_tsne = tsne.fit_transform(X_scaled)

customer_profiles['tsne_1'] = X_tsne[:, 0]
customer_profiles['tsne_2'] = X_tsne[:, 1]
print("t-SNE complete!")

# Create visualizations
fig, axes = plt.subplots(2, 3, figsize=(20, 12))

# 1. GMM Clusters
scatter1 = axes[0, 0].scatter(customer_profiles['tsne_1'], 
                              customer_profiles['tsne_2'],
                              c=customer_profiles['gmm_cluster'], 
                              cmap='tab10', 
                              alpha=0.6, s=50)
axes[0, 0].set_title('t-SNE: GMM Clusters')
axes[0, 0].set_xlabel('t-SNE Component 1')
axes[0, 0].set_ylabel('t-SNE Component 2')
plt.colorbar(scatter1, ax=axes[0, 0], label='Cluster')

# 2. Confidence
scatter2 = axes[0, 1].scatter(customer_profiles['tsne_1'], 
                              customer_profiles['tsne_2'],
                              c=customer_profiles['gmm_confidence'], 
                              cmap='RdYlGn', 
                              alpha=0.6, s=50, vmin=0, vmax=1)
axes[0, 1].set_title('t-SNE: Cluster Confidence')
axes[0, 1].set_xlabel('t-SNE Component 1')
axes[0, 1].set_ylabel('t-SNE Component 2')
plt.colorbar(scatter2, ax=axes[0, 1], label='Confidence')

# 3. Income
scatter3 = axes[0, 2].scatter(customer_profiles['tsne_1'], 
                              customer_profiles['tsne_2'],
                              c=customer_profiles['avg_income'], 
                              cmap='viridis', 
                              alpha=0.6, s=50)
axes[0, 2].set_title('t-SNE: Income Levels')
axes[0, 2].set_xlabel('t-SNE Component 1')
axes[0, 2].set_ylabel('t-SNE Component 2')
plt.colorbar(scatter3, ax=axes[0, 2], label='Income')

# 4. Total Spend
scatter4 = axes[1, 0].scatter(customer_profiles['tsne_1'], 
                              customer_profiles['tsne_2'],
                              c=customer_profiles['customer_total_spend'], 
                              cmap='plasma', 
                              alpha=0.6, s=50)
axes[1, 0].set_title('t-SNE: Total Spending')
axes[1, 0].set_xlabel('t-SNE Component 1')
axes[1, 0].set_ylabel('t-SNE Component 2')
plt.colorbar(scatter4, ax=axes[1, 0], label='Spend')

# 5. Age Groups
if 'age_group' in customer_profiles.columns:
    age_map = {age: i for i, age in enumerate(customer_profiles['age_group'].unique())}
    scatter5 = axes[1, 1].scatter(customer_profiles['tsne_1'], 
                                  customer_profiles['tsne_2'],
                                  c=customer_profiles['age_group'].map(age_map), 
                                  cmap='Set3', 
                                  alpha=0.6, s=50)
    axes[1, 1].set_title('t-SNE: Age Groups')
    axes[1, 1].set_xlabel('t-SNE Component 1')
    axes[1, 1].set_ylabel('t-SNE Component 2')
    cbar5 = plt.colorbar(scatter5, ax=axes[1, 1])
    cbar5.set_ticks(range(len(age_map)))
    cbar5.set_ticklabels(list(age_map.keys()))

# 6. Cluster Sizes
cluster_sizes = customer_profiles['gmm_cluster'].value_counts().sort_index()
axes[1, 2].bar(cluster_sizes.index, cluster_sizes.values, color='steelblue', alpha=0.7)
axes[1, 2].set_title('Cluster Size Distribution')
axes[1, 2].set_xlabel('Cluster ID')
axes[1, 2].set_ylabel('Customers')
axes[1, 2].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('gmm_tsne_visualization.png', dpi=300, bbox_inches='tight')
plt.show()

# Analyze clusters
print("\n" + "="*70)
print("GMM CLUSTER ANALYSIS")
print("="*70)

for cluster_id in sorted(customer_profiles['gmm_cluster'].unique()):
    cluster_data = customer_profiles[customer_profiles['gmm_cluster'] == cluster_id]
    
    print(f"\nCluster {cluster_id}: {len(cluster_data)} customers ({len(cluster_data)/len(customer_profiles)*100:.1f}%)")
    print(f"  Avg Income: Rs.{cluster_data['avg_income'].mean():,.0f}")
    print(f"  Avg Spend: Rs.{cluster_data['customer_total_spend'].mean():,.0f}")
    print(f"  Avg Transactions: {cluster_data['customer_total_transactions'].mean():.1f}")
    print(f"  Avg Confidence: {cluster_data['gmm_confidence'].mean():.3f}")
    
    if 'age_group' in cluster_data.columns:
        print(f"  Primary Age: {cluster_data['age_group'].mode()[0]}")
    if 'city' in cluster_data.columns:
        print(f"  Primary City: {cluster_data['city'].mode()[0]}")

# Uncertain customers
uncertain = customer_profiles[customer_profiles['gmm_confidence'] < 0.7]
print(f"\n" + "="*70)
print(f"UNCERTAIN CUSTOMERS: {len(uncertain)} ({len(uncertain)/len(customer_profiles)*100:.1f}%)")
print("="*70)
print("These customers span multiple segments")

# Save results
customer_profiles.to_csv('customer_profiles_gmm_tsne.csv', index=False)
print(f"\nResults saved to 'customer_profiles_gmm_tsne.csv'")

con.close()
print("\nAnalysis complete!")
```
